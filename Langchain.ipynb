# 1. Imports for Shopping Chatbot


```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
llm = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key="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"
)
```

# 2. Initialize the Groq LLM


```python
shopping_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are an AI Shopping & Fashion Assistant.

Rules:
- Answer ONLY shopping and fashion related questions.
- Extract the brand, category and budget from the user's question automatically.
- If any detail is missing, make a reasonable assumption or politely ask for clarification.
- Always answer in exactly 6-7 bullet points.
- Start every bullet with •.
- Mention:
  • Product name
  • Estimated price
  • Colors
  • Material
  • Features
  • Best use
  • Recommendation
- Never write long paragraphs.
"""
    ),
    ("human", "{question}")
])
```

# 3. Create Shopping Prompt Template


```python
# Store user history in memory
history_store = {}

def get_history(session_id: str):
    if session_id not in history_store:
        history_store[session_id] = InMemoryChatMessageHistory()
    return history_store[session_id]

shopping_chain = RunnableWithMessageHistory(
    shopping_prompt | llm,
    get_history,
    input_messages_key="question",
    history_messages_key="history"
)
```

```python
# Store user history in memory
history_store = {}

def get_history(session_id: str):
    if session_id not in history_store:
        history_store[session_id] = InMemoryChatMessageHistory()
    return history_store[session_id]

shopping_chain = RunnableWithMessageHistory(
    shopping_prompt | llm,
    get_history,
    input_messages_key="question",
    history_messages_key="history"
)
```

# 4. Manage Conversation History and Test the Shopping Chatbot


```python
response = shopping_chain.invoke(
    {
        "brand": "Lee Cooper",
        "category": "Top",
        "price": "Under ₹4000",
        "question": "Recommend winter outfits for women."
    },
    config={
        "configurable":{
            "session_id":"user1"
        }
    }
)

print(response.content)
```

# 5. Build Gradio Interface


```python
import gradio as gr

def chat_with_bot(user_input, session_id="user1"):

    response = shopping_chain.invoke(
        {"question": user_input},
        config={
            "configurable": {
                "session_id": session_id
            }
        }
    )

    return response.content


demo = gr.Interface(
    fn=chat_with_bot,
    inputs=gr.Textbox(
        label="Ask Your Shopping Question",
        placeholder="Example: Suggest Nike T-shirts under ₹2000"
    ),
    outputs=gr.Markdown(),
    title="🛍️ Shopping Fashion Chatbot",
    description="Ask anything related to shopping and fashion."
)

demo.launch()
```
